# 01. Crie uma estrutura com try/except para converter temperatura celsius em fahreinht

```python
try:
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = (celsius * 9/5) + 32
    print(f"Temperature in Fahrenheit: {fahrenheit}")
except ValueError:
    print("Error: Please enter a valid number for temperature.")
```

# 2. Crie uma estrutura com try/except a qual calcula e traga um erro para calcular a area de um retantuglo e circulo com estrutura de input para o usuario. Alem disso, o usuario deve escolher a figura geometrica

```python
import math

try:
    shape = input("Choose a shape (rectangle/circle): ").lower()
    
    if shape == "rectangle":
        length = float(input("Enter the length: "))
        width = float(input("Enter the width: "))
        area = length * width
        print(f"Rectangle area: {area}")
    
    elif shape == "circle":
        radius = float(input("Enter the radius: "))
        area = math.pi * radius ** 2
        print(f"Circle area: {area}")
    
    else:
        print("Erro: Forma inválida. Por favor escolha 'retângulo' ou 'círculo'.")
        
except ValueError:
    print("Erro: Por favor insira um número válido")
```
